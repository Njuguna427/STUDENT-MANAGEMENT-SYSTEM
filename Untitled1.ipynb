```python
#write a python program that displays the largest number
def get_largest(a, b):
    return max(a, b)
result = get_largest(15, 42)
print("The largest number is:", result)
```

```python
#create a list of twenty numbers loop through the list and print the even numbers
numbers = list(range(1, 21))
print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```
